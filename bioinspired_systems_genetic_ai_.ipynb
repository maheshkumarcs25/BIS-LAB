```python
import random

# 1. Define the fitness function
def fitness(x):
    return x ** 2


# 2. Initialize GA parameters
POPULATION_SIZE = 10
MUTATION_RATE = 0.01
CROSSOVER_RATE = 0.8
GENERATIONS = 50
CHROMOSOME_LENGTH = 5


# 3. Create initial population
def create_population():
    return [
        [random.randint(0, 1) for _ in range(CHROMOSOME_LENGTH)]
        for _ in range(POPULATION_SIZE)
    ]


# Convert binary chromosome to decimal
def decode(chromosome):
    return int("".join(map(str, chromosome)), 2)


# 4. Evaluate fitness
def evaluate(population):
    return [fitness(decode(individual)) for individual in population]


# 5. Selection using tournament selection
def selection(population):
    selected = []

    for _ in range(POPULATION_SIZE):
        a, b = random.sample(population, 2)

        if fitness(decode(a)) > fitness(decode(b)):
            selected.append(a[:])
        else:
            selected.append(b[:])

    return selected


# 6. Crossover
def crossover(parent1, parent2):
    if random.random() < CROSSOVER_RATE:
        point = random.randint(1, CHROMOSOME_LENGTH - 1)

        child1 = parent1[:point] + parent2[point:]
        child2 = parent2[:point] + parent1[point:]

        return child1, child2

    return parent1[:], parent2[:]


# 7. Mutation
def mutate(chromosome):
    for i in range(CHROMOSOME_LENGTH):
        if random.random() < MUTATION_RATE:
            chromosome[i] = 1 - chromosome[i]

    return chromosome


# 8. Genetic Algorithm
population = create_population()

best_solution = None
best_fitness = float("-inf")

for generation in range(GENERATIONS):

    # Evaluate current population
    fitness_values = evaluate(population)

    # Track the best solution
    max_fitness = max(fitness_values)
    index = fitness_values.index(max_fitness)

    if max_fitness > best_fitness:
        best_fitness = max_fitness
        best_solution = population[index][:]

    # Selection
    selected = selection(population)

    # Crossover and mutation
    new_population = []

    for i in range(0, POPULATION_SIZE, 2):
        parent1 = selected[i]
        parent2 = selected[(i + 1) % POPULATION_SIZE]

        child1, child2 = crossover(parent1, parent2)

        child1 = mutate(child1)
        child2 = mutate(child2)

        new_population.extend([child1, child2])

    population = new_population[:POPULATION_SIZE]

    print(
        f"Generation {generation + 1}: "
        f"Best x = {decode(best_solution)}, "
        f"Fitness = {best_fitness}"
    )


# 9. Output the best solution
print("\nBest Solution Found:")
print("x =", decode(best_solution))
print("Maximum value =", best_fitness)
```
