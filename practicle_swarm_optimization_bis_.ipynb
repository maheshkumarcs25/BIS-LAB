```python
import random
import math

def fitness(x):
    return x * math.sin(x)

N = 20
W = 0.7
C1 = 1.5
C2 = 1.5
ITER = 50

X = [random.uniform(0, 10) for _ in range(N)]
V = [random.uniform(-1, 1) for _ in range(N)]

Pbest = X[:]
Pbest_value = [fitness(x) for x in X]

Gbest = Pbest[Pbest_value.index(max(Pbest_value))]
Gbest_value = fitness(Gbest)

for it in range(ITER):

    for j in range(N):

        value = fitness(X[j])

        if value > Pbest_value[j]:
            Pbest[j] = X[j]
            Pbest_value[j] = value

        if value > Gbest_value:
            Gbest = X[j]
            Gbest_value = value

        r1 = random.random()
        r2 = random.random()

        V[j] = (W * V[j]
                + C1 * r1 * (Pbest[j] - X[j])
                + C2 * r2 * (Gbest - X[j]))

        X[j] = X[j] + V[j]
        X[j] = max(0, min(10, X[j]))

    print("Iteration", it + 1,
          ": Best x =", round(Gbest, 4),
          ", Fitness =", round(Gbest_value, 4))

print("\nFinal Result")
print("Best x =", round(Gbest, 4))
print("Maximum value =", round(Gbest_value, 4))
```
