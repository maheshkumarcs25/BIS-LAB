```python
#Parallel Cellular Algorithms and Programs

import random
GRID_SIZE = 10
ITERATIONS = 5
MIN_VALUE = -10
MAX_VALUE = 10
grid = [
    [
        [random.uniform(MIN_VALUE, MAX_VALUE),
         random.uniform(MIN_VALUE, MAX_VALUE)]
        for _ in range(GRID_SIZE)
    ]
    for _ in range(GRID_SIZE)
]
def fitness(position):
    x, y = position
    return x * x + y * y
def get_neighbors(i, j):
    neighbors = []
    for di in [-1, 0, 1]:
        for dj in [-1, 0, 1]:
            if di == 0 and dj == 0:
                continue
            ni = i + di
            nj = j + dj
            if 0 <= ni < GRID_SIZE and 0 <= nj < GRID_SIZE:
                neighbors.append(grid[ni][nj])
    return neighbors
best_position = None
best_fitness = float("inf")
for iteration in range(ITERATIONS):
    new_grid = [[None for _ in range(GRID_SIZE)]
                for _ in range(GRID_SIZE)]
    for i in range(GRID_SIZE):
        for j in range(GRID_SIZE):
            current = grid[i][j]
            current_fitness = fitness(current)
            neighbors = get_neighbors(i, j)
            best_neighbor = current
            for neighbor in neighbors:
                if fitness(neighbor) < fitness(best_neighbor):
                    best_neighbor = neighbor
            x, y = current
            bx, by = best_neighbor
            new_x = x + 0.5 * (bx - x)
            new_y = y + 0.5 * (by - y)
            new_x += random.uniform(-0.1, 0.1)
            new_y += random.uniform(-0.1, 0.1)
            new_x = max(MIN_VALUE, min(MAX_VALUE, new_x))
            new_y = max(MIN_VALUE, min(MAX_VALUE, new_y))
            new_grid[i][j] = [new_x, new_y]
            new_fitness = fitness([new_x, new_y])
            if new_fitness < best_fitness:
                best_fitness = new_fitness
                best_position = [new_x, new_y]
    grid = new_grid
    print(
        f"Iteration {iteration + 1}: "
        f"Best = {best_position}, "
        f"Fitness = {best_fitness:.6f}"
    )
print("\nBest Solution:")
print("x =", best_position[0])
print("y =", best_position[1])
print("Fitness =", best_fitness)
```
